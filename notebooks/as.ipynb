```python
#yOUR CODES
null = clean_df.isnull().sum()
ratio = (null / clean_df.shape[0])*100
pd.DataFrame({'Null_sum':null,'Ratio':ratio}).T
```



> ## <mark>The result above shows many missing data </mark>



###  To fix this issue You will fill these values with ..median.. for numeric and ..mode..for categorical

```python
# trying to fill missing values in my datset using for loop 

for col in clean_df.columns:
    if clean_df[col].dtype in['float','int']:
        clean_df[col].fillna(clean_df[col].median())
    else:
        clean_df[col].fillna(clean_df[col].mode()[0])


```

# <a id="9">    9 ) . How long is the duration of the trip for each gender in minutes?  </a>

```python
# Make visualization to easy get insight

# ------------------------Duration distribution by member gender

avg_duration_gender = (clean_df
                        .groupby('member_gender', observed=False)['min_duration']
                        .mean()
                        .reset_index())

fig = px.bar(avg_duration_gender,
            x='member_gender',
            y='min_duration',
            title="<b>Average Trip Duration (Minutes) by Gender</b>",
            color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_layout(width=800, height=500)
fig.show()

```

# Make visualization to easy get insight


The trip distribution by gender shows that female users have a higher average trip duration in minutes compared to male users, while the “Other” category represents users with unknown or unspecified gender in the dataset.

# <a id="10 "> 10 ) . How long is the duration of the trip for each user_type in minutes?  </a>

```python

# Make visualization to easy get insight

avg_duration_user = (clean_df
                    .groupby('user_type', observed=False)['min_duration']
                    .mean()
                    .reset_index())

fig = px.bar(avg_duration_user,
            x='user_type',
            y='min_duration',
            title="<b>Average Trip Duration (Minutes) by User Type</b>",
            color_discrete_sequence=px.colors.qualitative.Pastel)

fig.update_layout(width=800,
                    height=500,
                    yaxis_title="Average Duration (minutes)",
                    xaxis_title="User Type")

fig.show()

```

the average distribution according to user type indicates that customers generally take longer trips, while subscribers tend to have shorter trips, showing that trip duration varies depending on the user type.

# <a id="14"> 14 )  .  What is the  Correlation between `specific numeric variables` in the dataset?</a>


```python
# Make visualization to easy get insight
import plotly.express as px

cols = ['duration_sec','min_duration','hour_duration','age','member_birth_year']
corr_matrix = clean_df[cols].corr()

fig = px.imshow(corr_matrix,
                text_auto=".2f",
                color_continuous_scale="RdBu",
                zmin=-1, zmax=1,
                title="Artificial Correlation Between Derived Numeric Variables",
                aspect="auto")  
fig.update_layout( width= 800, height=500)
fig.show()



```


- There is no significant correlation between trip duration and rider age or birth year.

- Strong correlations between duration_sec, duration_min, and duration_hour are expected because they represent the same trip duration in different units.

- A strong negative correlation between age and member_birth_year occurs because age is derived from birth year.




# <a id="15">  15 ) What is the  Correlation between data_stat variables in the dataset?</a>


```python

"""""This time I will use heatmap  to explain how Correlation between data_stat variables

in the dataset in more clear and specific percentage"""

import plotly.express as px

data_stat = clean_df[['member_gender',
                    'user_type',
                    'bike_share_for_all_trip']]

data_stat_encoded = data_stat.apply(
    lambda x: x.astype('category').cat.codes
)

corr_matrix = data_stat_encoded.corr()

fig = px.imshow(corr_matrix,
                text_auto=".2f",
                color_continuous_scale="RdBu",
                zmin=-1,
                zmax=1,
                title="Correlation Between Data_stat Variables",
                aspect="auto")
fig.show()


```

The heatmap helps us easily see how the numeric variables are related to each other.

- duration_sec, duration_min, and duration_hour are strongly correlated, which makes sense because they all describe the same trip duration using different units.

- age and member_birth_year show a strong negative correlation, which is normal since age is calculated from birth year.

- Trip duration does not show a strong relationship with rider age, meaning people of different ages tend to have similar trip lengths.

- Variables like bike_id do not show meaningful correlation because they are just identification numbers, not real numerical measures.
