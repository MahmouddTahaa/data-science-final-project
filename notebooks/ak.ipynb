Function of info. of data

```python
def get_styled_info(df):
    """
    get info. of data in table includes data type, non-null count,
    null count, % missing, and unique values.
    """
    info_df = pd.DataFrame(index=df.columns)
    info_df['Data Type'] = df.dtypes
    info_df['Total Count'] = len(df)
    info_df['Non-Null Count'] = df.count()
    info_df['Null Count'] = df.isnull().sum()
    info_df['% Missing'] = (df.isnull().sum() / len(df)) * 100
    info_df['Unique Values'] = df.nunique()

    styled_df = info_df.style\
        .bar(subset=['% Missing'], color='#d65f5f', vmin=0, vmax=100)\
        .background_gradient(cmap='Blues', subset=['Unique Values'])\
        .background_gradient(cmap='Greens', subset=['Non-Null Count'])\
        .format({
            '% Missing': '{:.01f}%',
            'Non-Null Count': '{:,}',
            'Null Count': '{:,}',
            'Unique Values': '{:,}'
        })\
        .set_caption("📊 Dataframe Summary & Quality Check")\
        .set_properties(**{'border-color': 'black', 'border-style': 'solid'})

    return styled_df

get_styled_info(df)
```

###  To fix this issue You will fill these values with mean for numeric and mod for categorical

```python
# trying to fill missing values in my datset using for loop

# Convert IDs to string so they are treated as Categorical, not Numeric
df['start_station_id'] = df['start_station_id'].astype(str)
df['end_station_id'] = df['end_station_id'].astype(str)

for col in df.columns:

    # Check if the column actually has missing values (to save time)
    if df[col].isnull().sum() > 0:

        # 1. If the column is Numeric (Float or Int) -> Fill with MEAN
        if pd.api.types.is_numeric_dtype(df[col]):
            mean_value = df[col].mean()
            df[col] = df[col].fillna(mean_value)
            print(f"Filled numeric column '{col}' with Mean: {mean_value:.2f}")

        # 2. If the column is Categorical (Object/String) -> Fill with MODE
        else:
            # .mode() returns a Series (list), so we take [0] to get the first value
            mode_value = df[col].mode()[0]
            df[col] = df[col].fillna(mode_value)
            print(f"Filled categorical column '{col}' with Mode: {mode_value}")
```
